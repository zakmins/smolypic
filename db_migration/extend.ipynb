```python
from pathlib import Path
import sqlite3

db_path = Path('./data/smolympic.db')
if not db_path.exists():
    alt = Path.cwd() / 'db_migration' / 'data' / 'smolympic.db'
    if alt.exists():
        db_path = alt
    else:
        raise FileNotFoundError(f'DB not found: {db_path}')

con = sqlite3.connect(str(db_path))
cur = con.cursor()

start_date = '2026-10-04'
end_date = '2026-10-16'
days_to_extend = 13

affected = cur.execute("""
    SELECT id, name, sub_start, sub_end
    FROM members
    WHERE membership_type = 'subscription'
      AND date(sub_end) >= date(:start_date)
      AND date(sub_start) <= date(:end_date)
    ORDER BY sub_end
""", {'start_date': start_date, 'end_date': end_date}).fetchall()

print(f'Rows to update: {len(affected)}')
for item in affected:
    print(item)

cur.execute("""
    UPDATE members
    SET sub_end = datetime(sub_end, '+' || :days_to_extend || ' days')
    WHERE membership_type = 'subscription'
      AND date(sub_end) >= date(:start_date)
      AND date(sub_start) <= date(:end_date)
""", {'days_to_extend': days_to_extend, 'start_date': start_date, 'end_date': end_date})

con.commit()
print('Updated sub_end only. Sessions were NOT changed.')
con.close()
```
